```python
# Convert Purchase Date to datetime
large['Purchase Date'] = pd.to_datetime(large['Purchase Date'])

# Check the date range
print("Minimum date:", large['Purchase Date'].min())
print("Maximum date:", large['Purchase Date'].max())
```

```python
# Check duplicate rows
print("Duplicate rows:", large.duplicated().sum())
```

```python
# Remove duplicate rows
large = large.drop_duplicates()

# Check missing values
print("Missing values after removing duplicates:")
print(large.isnull().sum())

# Check data types
print("\nData types:")
print(large.dtypes)
```

```python
# Set the analysis date as the day after the latest purchase
analysis_date = large['Purchase Date'].max() + pd.Timedelta(days=1)

# Create RFM table
rfm = large.groupby('Customer ID').agg(
    Recency=('Purchase Date', lambda x: (analysis_date - x.max()).days),
    Frequency=('Purchase Date', 'count'),
    Monetary=('Total Purchase Amount', 'sum')
).reset_index()

# Display the RFM data
rfm.head()
```

```python
# Create RFM scores from 1 to 5

rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

# Convert scores to numbers
rfm[['R_Score', 'F_Score', 'M_Score']] = rfm[
    ['R_Score', 'F_Score', 'M_Score']
].astype(int)

# Create total RFM score
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

rfm.head()
```

```python
def segment_customer(score):
    if score >= 13:
        return "Champions"
    elif score >= 10:
        return "Loyal Customers"
    elif score >= 7:
        return "Potential Customers"
    elif score >= 5:
        return "At Risk"
    else:
        return "Lost Customers"

rfm['Segment'] = rfm['RFM_Score'].apply(segment_customer)

# See the segments
print(rfm['Segment'].value_counts())

# Show sample
rfm.head()
```

```python
import matplotlib.pyplot as plt

segment_counts = rfm['Segment'].value_counts()

plt.figure(figsize=(8, 5))
segment_counts.plot(kind='bar')

plt.title('Customer Segmentation')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
segment_analysis = rfm.groupby('Segment').agg(
    Customers=('Customer ID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).round(2)

segment_analysis
```

```python
# Create monthly purchase data
large['Month'] = large['Purchase Date'].dt.to_period('M').astype(str)

monthly_sales = large.groupby('Month')['Total Purchase Amount'].sum()

plt.figure(figsize=(10, 5))
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Purchase Amount')
plt.xlabel('Month')
plt.ylabel('Total Purchase Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Check churn distribution
churn_counts = large['Churn'].value_counts()

print(churn_counts)

# Churn visualization
plt.figure(figsize=(6, 5))
churn_counts.plot(kind='bar')

plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Calculate churn rate
churn_rate = large['Churn'].mean() * 100

print(f"Overall Churn Rate: {churn_rate:.2f}%")
```

```python
# Churn rate by customer segment
segment_churn = large.merge(
    rfm[['Customer ID', 'Segment']],
    on='Customer ID',
    how='left'
).groupby('Segment')['Churn'].mean().mul(100).round(2)

print("Churn Rate by Segment:")
print(segment_churn)
```

```python
# Prepare churn data with customer segments
customer_data = large.merge(
    rfm[['Customer ID', 'Segment']],
    on='Customer ID',
    how='left'
)

# Calculate churn rate for each segment
segment_churn = (
    customer_data.groupby('Segment')['Churn']
    .mean()
    .mul(100)
    .round(2)
)

print("Churn Rate by Customer Segment:")
print(segment_churn)
```

```python
plt.figure(figsize=(8, 5))
segment_churn.plot(kind='bar')

plt.title('Churn Rate by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
    Image, PageBreak
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch
import matplotlib.pyplot as plt
import os

# -----------------------------
# Create charts for the report
# -----------------------------

# 1. Customer Segmentation
plt.figure(figsize=(8,5))
rfm['Segment'].value_counts().plot(kind='bar')
plt.title('Customer Segmentation')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.tight_layout()
segment_chart = '/content/segment_chart.png'
plt.savefig(segment_chart, dpi=150)
plt.close()

# 2. Monthly Purchase Trend
plt.figure(figsize=(9,5))
monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Purchase Amount')
plt.xlabel('Month')
plt.ylabel('Total Purchase Amount')
plt.xticks(rotation=45)
plt.tight_layout()
monthly_chart = '/content/monthly_purchase.png'
plt.savefig(monthly_chart, dpi=150)
plt.close()

# 3. Churn Distribution
plt.figure(figsize=(7,5))
large['Churn'].value_counts().plot(kind='bar')
plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.tight_layout()
churn_chart = '/content/churn_distribution.png'
plt.savefig(churn_chart, dpi=150)
plt.close()

# 4. Churn by Segment
plt.figure(figsize=(8,5))
segment_churn.plot(kind='bar')
plt.title('Churn Rate by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
segment_churn_chart = '/content/churn_by_segment.png'
plt.savefig(segment_churn_chart, dpi=150)
plt.close()


# -----------------------------
# Create PDF
# -----------------------------

pdf_path = '/content/Ecommerce_Customer_Segmentation_Report.pdf'

doc = SimpleDocTemplate(
    pdf_path,
    pagesize=A4,
    rightMargin=40,
    leftMargin=40,
    topMargin=40,
    bottomMargin=40
)

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    'TitleStyle',
    parent=styles['Title'],
    alignment=TA_CENTER,
    fontSize=20,
    spaceAfter=20
)

heading_style = ParagraphStyle(
    'HeadingStyle',
    parent=styles['Heading2'],
    fontSize=14,
    spaceBefore=12,
    spaceAfter=8
)

body_style = ParagraphStyle(
    'BodyStyle',
    parent=styles['BodyText'],
    fontSize=10,
    leading=15,
    spaceAfter=8
)

story = []

# Title
story.append(Paragraph(
    'E-Commerce Customer Segmentation & Retention Analysis',
    title_style
))

story.append(Paragraph(
    '<b>Customer Analytics Project</b>',
    ParagraphStyle(
        'sub',
        parent=body_style,
        alignment=TA_CENTER
    )
))

story.append(Spacer(1, 20))

# Introduction
story.append(Paragraph('1. Introduction', heading_style))
story.append(Paragraph(
    'This project analyzes e-commerce customer purchasing behavior '
    'to identify customer segments and understand customer retention '
    'and churn patterns. RFM analysis is used to classify customers '
    'according to their purchasing behavior.',
    body_style
))

# Objectives
story.append(Paragraph('2. Objectives', heading_style))

objectives = [
    'Clean and prepare the e-commerce dataset.',
    'Analyze customer purchasing behavior.',
    'Perform RFM analysis.',
    'Segment customers based on RFM scores.',
    'Analyze purchase trends and customer churn.',
    'Provide actionable business recommendations.'
]

for item in objectives:
    story.append(Paragraph('• ' + item, body_style))

# Dataset
story.append(Paragraph('3. Dataset Description', heading_style))
story.append(Paragraph(
    'The dataset contains customer and transaction information including '
    'Customer ID, Purchase Date, Product Category, Product Price, Quantity, '
    'Total Purchase Amount, Payment Method, Returns, Age, Gender and Churn.',
    body_style
))

# Data Cleaning
story.append(Paragraph('4. Data Cleaning', heading_style))
story.append(Paragraph(
    'The Purchase Date column was converted into datetime format. '
    'Missing values and duplicate records were checked, duplicate records '
    'were removed, and the data types of the columns were verified. '
    'A Month column was also created for monthly purchase analysis.',
    body_style
))

# RFM
story.append(Paragraph('5. RFM Analysis', heading_style))
story.append(Paragraph(
    '<b>Recency:</b> Number of days since the customer last purchased.<br/>'
    '<b>Frequency:</b> Number of purchases made by the customer.<br/>'
    '<b>Monetary:</b> Total amount spent by the customer.',
    body_style
))

story.append(Paragraph(
    'RFM scores ranging from 1 to 5 were calculated for each customer.',
    body_style
))

# Segmentation
story.append(Paragraph('6. Customer Segmentation', heading_style))
story.append(Paragraph(
    'Customers were classified into five groups based on their RFM scores: '
    'Champions, Loyal Customers, Potential Customers, At Risk, and Lost Customers.',
    body_style
))

story.append(Image(segment_chart, width=6.5*inch, height=4*inch))
story.append(Spacer(1, 10))

# Segment table
story.append(Paragraph('7. Segment Analysis', heading_style))

table_data = [
    ['Segment', 'Customers', 'Avg Recency', 'Avg Frequency', 'Avg Monetary']
]

for index, row in final_summary.iterrows():
    table_data.append([
        str(index),
        str(row['Customers']),
        f"{row['Avg_Recency']:.2f}",
        f"{row['Avg_Frequency']:.2f}",
        f"{row['Avg_Monetary']:.2f}"
    ])

table = Table(table_data, repeatRows=1)

table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.lightgrey),
    ('TEXTCOLOR', (0,0), (-1,0), colors.black),
    ('GRID', (0,0), (-1,-1), 0.5, colors.grey),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,-1), 8),
    ('ALIGN', (1,1), (-1,-1), 'CENTER'),
    ('VALIGN', (0,0), (-1,-1), 'MIDDLE')
]))

story.append(table)

# Purchase trend
story.append(PageBreak())
story.append(Paragraph('8. Purchase Trend Analysis', heading_style))
story.append(Paragraph(
    'Monthly purchase amounts were analyzed to identify changes in '
    'customer purchasing activity over time.',
    body_style
))
story.append(Image(monthly_chart, width=6.5*inch, height=3.7*inch))

# Churn
story.append(Paragraph('9. Churn Analysis', heading_style))
story.append(Paragraph(
    f'The calculated overall churn rate is '
    f'<b>{churn_rate:.2f}%</b>. Churn rates were also compared across '
    f'customer segments to identify groups requiring retention efforts.',
    body_style
))

story.append(Image(churn_chart, width=5.5*inch, height=3.7*inch))
story.append(Spacer(1, 10))
story.append(Image(segment_churn_chart, width=6.5*inch, height=3.7*inch))

# Recommendations
story.append(PageBreak())
story.append(Paragraph('10. Actionable Recommendations', heading_style))

recommendations = [
    '<b>Retain Champions:</b> Provide loyalty rewards, exclusive offers, and early access to new products.',
    '<b>Engage Loyal Customers:</b> Use personalized recommendations and targeted promotions to increase purchase frequency.',
    '<b>Reactivate At-Risk Customers:</b> Send win-back campaigns, discounts, and personalized offers.',
    '<b>Convert Potential Customers:</b> Encourage repeat purchases through personalized recommendations and limited-time offers.',
    '<b>Reduce Customer Churn:</b> Identify customers showing declining engagement and launch targeted retention campaigns.'
]

for item in recommendations:
    story.append(Paragraph('• ' + item, body_style))

# Conclusion
story.append(Paragraph('11. Conclusion', heading_style))
story.append(Paragraph(
    'The analysis demonstrates how e-commerce transaction data can be '
    'used to understand customer purchasing behavior. RFM segmentation, '
    'purchase trend analysis, and churn analysis provide useful insights '
    'for developing targeted marketing and customer retention strategies.',
    body_style
))

# Build PDF
doc.build(story)

print("PDF created successfully!")
print(pdf_path)
```
